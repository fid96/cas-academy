## 1. Imports et chargement

```python
import pandas as pd
import matplotlib.pyplot as plt

ventes = pd.read_csv("../data/ventes.csv")
clients = pd.read_csv("../data/clients.csv")

print("Ventes:", ventes.shape)
print("Clients:", clients.shape)
ventes.head()
```

## 2. Contrôle qualité

```python
print("=== Manquants ===")
print(ventes.isna().sum())

print("\n=== Types ===")
print(ventes.dtypes)

print("\n=== Doublons ===")
print("Lignes en double:", ventes.duplicated().sum())

print("\n=== Villes présentes ===")
print(ventes["ville"].unique())
```

```python
# Préparation propre pour l'analyse
df = ventes.copy()
df["date"] = pd.to_datetime(df["date"])

# Documentez ce choix : on retire les lignes sans quantité pour les KPI quantité
df_q = df.dropna(subset=["quantite"]).copy()
df_q["quantite"] = df_q["quantite"].astype(int)

print("Lignes avant:", len(df), "| après dropna(quantite):", len(df_q))
```

## 3. Jointure clients + KPI

```python
analyse = df.merge(clients, on="client_id", how="left")

print("Total CDF:", analyse["montant_cdf"].sum())
print("\n--- Par ville ---")
print(analyse.groupby("ville")["montant_cdf"].sum().sort_values(ascending=False))
print("\n--- Par produit ---")
print(analyse.groupby("produit")["montant_cdf"].sum().sort_values(ascending=False))
print("\n--- Par type_client ---")
print(analyse.groupby("type_client")["montant_cdf"].sum().sort_values(ascending=False))
```

## 4. Tendance mensuelle

```python
analyse["mois"] = analyse["date"].dt.to_period("M").astype(str)
mensuel = analyse.groupby("mois")["montant_cdf"].sum()
print(mensuel)

mensuel.plot(kind="line", marker="o", color="#2d6a4f")
plt.title("Évolution mensuelle des ventes (CDF)")
plt.xlabel("Mois")
plt.ylabel("Montant (CDF)")
plt.tight_layout()
plt.show()
```

## 5. Graphique décisionnel — totaux par ville

```python
totaux_ville = analyse.groupby("ville")["montant_cdf"].sum().sort_values(ascending=False)

totaux_ville.plot(kind="bar", color="#3d7ea6")
plt.title("Total des ventes par ville (CDF)")
plt.xlabel("Ville")
plt.ylabel("Montant (CDF)")
plt.tight_layout()
plt.show()
```
